# Water Potability Analysis
All living beings need water. Water is a natural resource that exists in many forms. It exists in oceans, streams, rivers, lakes, and more.

Water that is safe to drink is defined as potable water. Scientists have studied water for many generations analyzing their chemical and physical properties, as well as which properties can be measured to determine the potability of a body of water.

The Water Potability Analysis is software which determines whether a body of water is potable for humans. It does this with a GUI frontend, for the user to upload photos, videos, or a live-stream with a body of water. The frontend will send this data to the backend.

The backend analyzes the input, and returns a decision to the frontend for a user.
